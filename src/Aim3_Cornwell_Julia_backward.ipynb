```julia
# Import necessary packages
using DynLoco_TC, Plots, CSV, DataFrames, Statistics, StructArrays

# Constants
g = 9.81
L = 1
norm_t = (L/g)^0.5
norm_v = (g*L)^0.5
nsteps = 7
pert_mag = 0.5
subj = 3

if subj == 0
    speed = 1.15/norm_v # average
elseif subj == 1
    speed = 1.05/norm_v
elseif subj == 2
    speed = 1.15/norm_v
elseif subj == 3
    speed = 1.25/norm_v
end

# Clear variables
wstar = nothing

# Nominal gait
wstar = findgait(WalkRW2lvs(safety=true,beta=0.49), target=:speed=>speed, varying=:P); # varies SL nonlinearly
nom_u = optwalk_TC(wstar, nsteps, boundaryvels = (wstar.vm, wstar.vm), boundarywork=false, J="u");

# Perturbed gait
p_u = optwalk_TC(wstar, nsteps, boundaryvels = (wstar.vm, wstar.vm), boundarywork=false, J="u", perts = [1 1 pert_mag 1 1 1 1]);
p_dv = optwalk_TC(wstar, nsteps, boundaryvels = (wstar.vm, wstar.vm), boundarywork=false, J="dv", perts = [1 1 pert_mag 1 1 1 1]);
```

```julia
using MAT
if subj == 0
    matwrite("proactive_u.mat", Dict("proactive_u" => p_u))
elseif subj == 1
    matwrite("S1_proactive_u.mat", Dict("proactive_u" => p_u))
elseif subj == 2
    matwrite("S2_proactive_u.mat", Dict("proactive_u" => p_u))
elseif subj == 3
    matwrite("S3_proactive_u.mat", Dict("proactive_u" => p_u))
end
```

```julia
using MAT
if subj == 0
    matwrite("proactive_dv.mat", Dict("proactive_dv" => p_dv))
elseif subj == 1
    matwrite("S1_proactive_dv.mat", Dict("proactive_dv" => p_dv))
elseif subj == 2
    matwrite("S2_proactive_dv.mat", Dict("proactive_dv" => p_dv))
elseif subj == 3
    matwrite("S3_proactive_dv.mat", Dict("proactive_dv" => p_dv))
end
```

```julia
using MAT
if subj == 0
    matwrite("nominal_u.mat", Dict("nominal_u" => nom_u))
elseif subj == 1
    matwrite("S1_nominal_u.mat", Dict("nominal_u" => nom_u))
elseif subj == 2
    matwrite("S2_nominal_u.mat", Dict("nominal_u" => nom_u))
elseif subj == 3
    matwrite("S3_nominal_u.mat", Dict("nominal_u" => nom_u))
end
```

```julia
# Naive control: Apply nominal controls to perturbation condition
naive = multistep(wstar, Ps=nom_u.steps.P, perts=[1 1 pert_mag 1 1 1 1])

# Reactive control: No anticipatory control - just react after the perturbation happens
reactive_1 = multistep(wstar, Ps=nom_u.steps.P[1:3], perts = [1 1 pert_mag])
reactive_2u = optwalk_TC(wstar, 4, boundaryvels=(reactive_1.steps[end].vm,wstar.vm), boundarywork=false, perts=[1 1 1 1], J="u", avg_speed=0)
reactive_u = cat(reactive_1, reactive_2u);
```

```julia
using MAT
if subj == 0
    matwrite("naive.mat", Dict("naive" => naive))
elseif subj == 1
    matwrite("S1_naive.mat", Dict("naive" => naive))
elseif subj == 2
    matwrite("S2_naive.mat", Dict("naive" => naive))
elseif subj == 3
    matwrite("S3_naive.mat", Dict("naive" => naive))
end
```

```julia
using MAT
if subj == 0
    matwrite("reactive_u.mat", Dict("reactive_u" => reactive_u))
elseif subj == 1
    matwrite("S1_reactive_u.mat", Dict("reactive_u" => reactive_u))
elseif subj == 2
    matwrite("S2_reactive_u.mat", Dict("reactive_u" => reactive_u))
elseif subj == 3
    matwrite("S3_reactive_u.mat", Dict("reactive_u" => reactive_u))
end
```
