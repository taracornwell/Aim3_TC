```julia
using DynLoco_TC; Plots; 

# Create structure
wstar = findgait(WalkRW2l(α=0.4,safety=true), target=:speed=>0.5, varying=:P)

# Set number of steps: pre, pert, recovery
nsteps = 3

# Run optimization that minimizes push-off work
nominal_stepresult = optwalk(wstar, nsteps, boundarywork=false)

println("Push-off work cost = ", nominal_stepresult.totalcost)

# Plots
p_nom = multistepplot(nominal_stepresult, boundarywork=false, legend=false) # plot speed, push-off
display(p_nom)

p_nom = plot(cumsum(nominal_stepresult.steps.tf), nominal_stepresult.steps.vm, xlabel="time", ylabel="midstance speed", legend=false)
display(p_nom)

plot(cumsum(nominal_stepresult.steps.tf), nominal_stepresult.steps.tf, xlabel="time", ylabel="step time", legend=false)
```
